```python
import os
import cv2


labels_path = 'labels_studio'
img_path = 'imgs'
label_destiny = 'labels'


lista_labels = sorted(os.listdir(labels_path))
lista_img    = sorted(os.listdir(img_path))
idx = 0
for i in range(len(lista_img)):
    if 2*i +1 > 60:
        break
    cm1 =  os.path.join(labels_path, lista_labels[2*i] )
    cm2 = os.path.join(labels_path, lista_labels[2*i+1])
    
    img1 = cv2.imread(cm1)
    img2 = cv2.imread(cm2)

    nome, ext = os.path.splitext(lista_img[i])
    nome1 = f'{nome}_BF.jpg'
    nome2 = f'{nome}_GM.jpg'

    save1 = os.path.join(label_destiny, nome1)
    save2 = os.path.join(label_destiny, nome2)

    cv2.imwrite(save1, img1)
    cv2.imwrite(save2, img2)
```
