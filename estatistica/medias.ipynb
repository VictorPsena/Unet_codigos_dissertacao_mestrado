## Verificando se as medidas de Nelore e Caracu são semelhantes

### Bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### Obtendo as medidas

```python
df_Nelore = pd.read_excel('medidas/Nelore.ods', engine='odf')
df_Nelore.head()
```

```python
df_Caracu = pd.read_excel('medidas/Caracu.ods', engine='odf')
df_Caracu.head()
```

### Modificando o banco de dados

```python
x_caracu = [i for i in range(len(df_Caracu['EGL']))]
x_nelore = [i for i in range(len(df_Nelore['AOL']))]
print(len(x_nelore))
print(len(x_caracu))

nelore_aol = df_Nelore['AOL'].mean()
caracu_aol = df_Caracu['AOL'].mean()


nelore_egl = df_Nelore['EGL'].mean()
caracu_egl = df_Caracu['EGL'].mean()


plt.scatter(x_caracu, sorted(df_Caracu['AOL']), label='CARACU_AOL')
plt.scatter(x_nelore, sorted(df_Nelore['AOL']), label='NELORE_AOL')
plt.axhline(caracu_aol, color='blue', linestyle='--', linewidth=1.5,
            label=f'Média Caracu: {caracu_aol:.2f}')
plt.axhline(nelore_aol, color='orange', linestyle='--', linewidth=1.5,
            label=f'Média Nelore: {nelore_aol:.2f}')

plt.xticks(rotation=45, ha='right')
plt.title('Comparação AOL entre Caracu e Nelore(27 imagens)')
plt.grid()
plt.legend()
plt.savefig('aol_comparacao.png', dpi = 300, bbox_inches='tight')
plt.show()
```
